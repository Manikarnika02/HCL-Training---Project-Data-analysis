```python

# LIBRARY CIRCULATION DATA CLEANING

import pandas as pd
import numpy as np
from google.colab import files

# STEP 1: Upload CSV dataset
uploaded = files.upload()
file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

# STEP 2: Display first five records
print("First five records:")
print(df.head())

# STEP 3: Check dataset size
print("\nDataset Shape:")
print(df.shape)

# STEP 4: Display column names
print("\nColumn Names:")
print(df.columns)

# STEP 5: Check data types
print("\nData Types:")
print(df.dtypes)

# STEP 6: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# STEP 7: Check duplicate records
print("\nDuplicate Records:", df.duplicated().sum())
print("Duplicate Loan IDs:", df["Loan_ID"].duplicated().sum())

# STEP 8: Fill missing Borrower_Age with median
age_median = df["Borrower_Age"].median()
df["Borrower_Age"] = df["Borrower_Age"].fillna(age_median)

# STEP 9: Convert dates into datetime format
df["Checkout_Date"] = pd.to_datetime(df["Checkout_Date"])
df["Return_Date"] = pd.to_datetime(df["Return_Date"])

# STEP 10: Fill missing Return_Date
loan_median = df["Loan_Duration_Days"].median()

df["Return_Date"] = df["Return_Date"].fillna(
    df["Checkout_Date"] + pd.to_timedelta(loan_median, unit="D")
)

# STEP 11: Remove duplicate Loan_ID records
df = df.drop_duplicates(subset="Loan_ID", keep="first")

# STEP 12: Calculate loan duration in days
df["Loan_Duration_Days"] = (
    df["Return_Date"] - df["Checkout_Date"]
).dt.days

# STEP 13: Check missing values after cleaning
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# STEP 14: Check duplicate records after cleaning
print("\nDuplicate Records After Cleaning:")
print(df.duplicated().sum())

# STEP 15: Display cleaned dataset
print("\nCleaned Dataset:")
print(df.head())

print("\nFinal Dataset Shape:")
print(df.shape)

# STEP 16: Save cleaned CSV file
df.to_csv("library_circulation_cleaned.csv", index=False)

print("\nData Cleaning Completed Successfully!")

# STEP 17: Download cleaned file
files.download("library_circulation_cleaned.csv")
```
